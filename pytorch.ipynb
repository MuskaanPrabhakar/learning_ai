MSELoss (Mean Squared Error) — Common for predicting continuous numbers (like exam scores or house prices).
CrossEntropyLoss — Common for classification problems (like predicting whether an email is spam or not).

```python
class StudentModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(2, 4)
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(4, 1)

    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        return x


model = StudentModel()

student1 = torch.tensor([[6.0, 80.0]])
actual_score = torch.tensor([[90.0]])

loss_fn = nn.MSELoss()

prediction = model(student1)

loss = loss_fn(prediction, actual_score)

print("Prediction:", prediction)
print("Loss:", loss)
print(model.layer1.weight.grad)
loss.backward()
print(model.layer1.weight.grad)
```

```python
print(model.layer1.bias.grad)
```

```python
import torch.optim as optim
class StudentModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(2, 4)

        self.relu = nn.ReLU()

        self.layer2 = nn.Linear(4, 1)

    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        return x


model = StudentModel()


student1 = torch.tensor([[6.0, 80.0]])
actual_score = torch.tensor([[90.0]])


loss_fn = nn.MSELoss()


optimizer = optim.SGD(model.parameters(), lr=0.001) #updates weight


prediction = model(student1)


loss = loss_fn(prediction, actual_score)


optimizer.zero_grad()


loss.backward()


optimizer.step()


print("Prediction:", prediction)
print("Loss:", loss)
```

```python
model.parameters()
```

```python
for epoch in range(100000):
    prediction = model(student1)

    loss = loss_fn(prediction, actual_score)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()
```
